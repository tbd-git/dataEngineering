```python
# User and Employee classes showing public, protected (convention), and private (name-mangling) fields

class User:
    def __init__(self, username, account_no, balance):
        self.username = username         # public
        self._account_no = account_no    # protected by convention
        self.__balance = balance         # private (name-mangled)

    # public getter for the private balance
    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

class Employee(User):
    def __init__(self, username, account_no, balance, emp_id, role):
        super().__init__(username, account_no, balance)
        self.emp_id = emp_id             # public
        self._role = role                # protected by convention
        self.__access_level = "employee" # private (name-mangled)

    def get_access_level(self):
        return self.__access_level

# Instances:
# 1) An employee who is also a user
employee = Employee("alice", "ACC123", 1500.0, emp_id="E001", role="teller")

# 2) A regular user who is not an employee
customer = User("bob", "ACC999", 300.0)

# Access examples:
print("Employee username (public):", employee.username)
print("Employee account (protected, accessible):", employee._account_no)
print("Employee balance (via getter):", employee.get_balance())
# direct access to private (not recommended) via name-mangling:
print("Employee balance (name-mangled):", employee._User__balance)

print("Employee id (public):", employee.emp_id)
print("Employee access level (via getter):", employee.get_access_level())
print("Employee access level (name-mangled):", employee._Employee__access_level)

print("Customer username (public):", customer.username)
print("Customer balance (via getter):", customer.get_balance())

# modify public field
customer.username = "robert"
print("Customer new username:", customer.username)
```

```python
import boto3
import csv
import io
from botocore.exceptions import ClientError

s3_client = boto3.client('s3')

bucket_name = 'aws-learning-omm'
file_key = 'test_aws_omm.csv'

try:
    response = s3_client.get_object(Bucket=bucket_name, Key=file_key)
    csv_content = response['Body'].read().decode('utf-8')
    text_stream = io.StringIO(csv_content)

    reader = csv.DictReader(text_stream)

    for row in reader:
        print(row)

except ClientError as e:
    error_code = e.response['Error']['Code']
    if error_code == 'NoSuchBucket':
        print(f"Error: Bucket '{bucket_name}' does not exist.")
    elif error_code == 'NoSuchKey':
        print(f"Error: File key '{file_key}' was not found in bucket.")
    elif error_code == 'AccessDenied':
        print("Error: Access denied. Check your AWS credentials or IAM permissions.")
    else:
        print(f"AWS S3 Error: {e}")
```
