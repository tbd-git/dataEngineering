#### 2-Cumulative Salary Calculation Example using Window Function:

```python
# 
from pyspark.sql import Window
import pyspark.sql.functions  as F
dataEmp = [
    (1, 'A', 'HR', 1000, 'Main Office', 'Expert', 'Full Time', "Bachelor's Degree", 'LinkedIn', 'N/A', 'Female', 'Single'),
    (2, 'B', 'IT', 3000, 'North Office', 'Vice-Director', 'Full Time', "Bachelor's Degree", 'Website', 'N/A', 'Male', 'Married'),
    (3, 'C', 'IT', 4000, 'East Office', 'Director', 'Part Time', 'High School', 'In-House', 'Resignation', 'Female', 'Divorced'),
    (4, 'D', 'Operations', 5000, 'Main Office', 'Director', 'Full Time', "Bachelor's Degree", 'LinkedIn', 'N/A', 'Male', 'Married'),
    (5, 'AD', 'HR', 1500, 'North Office', 'Expert', 'Home Office', "Bachelor's Degree", 'In-House', 'N/A', 'Female', 'Single'),
    (6, 'BD', 'IT', 3001, 'East Office', 'Vice-Director', 'Full Time', 'Primary School', 'Website', 'Retirement', 'Male', 'Married'),
    (7, 'CD', 'IT', 4100, 'Main Office', 'Expert', 'Home Office', "Bachelor's Degree", 'LinkedIn', 'N/A', 'Not specified', 'Single'),
    (8, 'DD', 'Operations', 5000, 'North Office', 'Director', 'Full Time', "Bachelor's Degree", 'In-House', 'N/A', 'Male', 'Married'),
    (9, 'AD', 'HR', 1500, 'East Office', 'Expert', 'Part Time', 'High School', 'Website', 'New Job', 'Female', 'Divorced'),
    (10, 'BD', 'IT', 3001, 'Main Office', 'Vice-Director', 'Full Time', "Bachelor's Degree", 'LinkedIn', 'N/A', 'Male', 'Single'),
    (11, 'CD', 'IT', 4100, 'North Office', 'Expert', 'Full Time', "Bachelor's Degree", 'In-House', 'N/A', 'Not specified', 'Married'),
    (12, 'DD', 'Operations', 5000, 'East Office', 'Director', 'Home Office', "Bachelor's Degree", 'Website', 'N/A', 'Male', 'Divorced'),
    (13, 'E', 'Marketing', 4500, 'Main Office', 'Expert', 'Full Time', "Bachelor's Degree", 'LinkedIn', 'N/A', 'Female', 'Single'),
    (14, 'F', 'Finance', 5500, 'North Office', 'Vice-Director', 'Full Time', "Bachelor's Degree", 'In-House', 'N/A', 'Male', 'Married'),
    (15, 'G', 'Finance', 6000, 'East Office', 'Director', 'Part Time', "Bachelor's Degree", 'Website', 'Resignation', 'Female', 'Divorced')
]
schemaEmp=('ID','Name','Department','Salary','Office','Job_Title','Job_Status','Education','Hiring_Source','Termination_Reason','Gender','Marital_Status')
dfEmp=spark.createDataFrame(dataEmp,schemaEmp)
windowSpec=Window.orderBy("ID")
dfCumSal=dfEmp.withColumn("cumulativeSalary",F.sum("salary").over(windowSpec))
dfCumSal.select("ID","Department","salary","cumulativeSalary").show()
```

#### 3-Get Highest Salary by Department

```python
departmentHighestSalary=Window.partitionBy('Department').orderBy(F.desc('salary'))
dfResult=dfEmp.withColumn('DepartmentHighestSalary',F.dense_rank().over(departmentHighestSalary))
dfHighestSalary=dfResult.filter(F.col("DepartmentHighestSalary")==1).show()
```
